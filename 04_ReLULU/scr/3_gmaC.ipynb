```python
import pandas as pd
from typing import List
```

```python
df = pd.read_excel("data/two_jobs_applications.xlsx")
df = df[df["Item"].str.contains("Ability C")]
```

```python
sequence_list = list(set([s.split("number? ")[1] for s in df.Question]))
```

```python
def get_second_difference(s_list):
    diff_list = []
    for i in range(1, len(s_list)):
        diff_list.append(s_list[i] - s_list[i - 1])
    return diff_list
```

```python
sequence = {}
for s in sequence_list:
    s_list = s.split(", ")[:-1]
    s_list = [int(n) for n in s_list]
    if (
        s_list[0] + s_list[1] == s_list[2] and
        s_list[1] + s_list[2] == s_list[3] and
        s_list[2] + s_list[3] == s_list[4]
    ):
        sequence[s] = ("fibonacci", s_list)
    else:
        diff_list = get_second_difference(s_list)
        if (
            diff_list[0] == diff_list[1] and
            diff_list[1] == diff_list[2]
        ):
            sequence[s] = ("arithmetic", s_list)
        elif (
            0 not in s_list and
            int(s_list[1]) % int(s_list[0]) == 0 and
            int(s_list[1]) / int(s_list[0]) == int(s_list[2]) / int(s_list[1]) and
            int(s_list[2]) / int(s_list[1]) == int(s_list[3]) / int(s_list[2])
        ):
            sequence[s] = ("geometric", s_list)
        elif (
            diff_list[1] - diff_list[0] == diff_list[2] - diff_list[1] and
            diff_list[2] - diff_list[1] == diff_list[3] - diff_list[2]
        ):
            sequence[s] = ("quadratic", s_list)
        elif (
            0 not in diff_list and
            diff_list[1] % diff_list[0] == 0 and
            diff_list[1] / diff_list[0] == diff_list[2] / diff_list[1]
        ):
            sequence[s] = ("second_geometric", s_list)
        else:
            sequence[s] = ("alternating", s_list)
```

```python
def solve_fibonacci(s: List[int]):
    next_num = s[-1] + s[-2]
    return (next_num, s[-1] + next_num)
```

```python
def solve_arithmetic(s: List[int]):
    diff = s[1] - s[0]
    return s[-1] + diff
```

```python
def solve_geometric(s: List[int]):
    ratio = s[1] / s[0]
    return int(s[-1] * ratio)
```

```python
def second_geometric(s: List[int]):
    d = get_second_difference(s)
    ratio = d[1] / d[0]
    return int(s[-1] + d[-1] * ratio)
```

```python
def solve_alternating(s: List[int]):
    new_s = [s[0], s[2], s[4]]
    return solve_arithmetic(new_s)
```

```python
answer = {}
for s_str, (sequence_type, s) in sequence.items():
    a = ""
    if sequence_type == "fibonacci":
        a = solve_fibonacci(s)
    if sequence_type == "arithmetic":
        a = solve_arithmetic(s)
    if sequence_type == "geometric":
        a = solve_geometric(s)
    if sequence_type == "second_geometric":
        a = second_geometric(s)
    if sequence_type == "alternating":
        a = solve_alternating(s)
    answer[s_str] = a
```

```python
response = [answer[s.split("number? ")[1]] for s in df.Question]
```

```python
df["Response"] = response
```

```python
df.to_csv("intermediate_data/gmaC.csv", index=False)
```
